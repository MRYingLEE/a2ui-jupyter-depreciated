# Button action

Run the first cell, then click **Click me**. The callback updates the status. Run the last cell to inspect the action.

```javascript
const { A2UISurface } = await import('/files/a2ui/kernel.js');
globalThis.surface = new A2UISurface();
surface.send([
  {
    "version": "1.0",
    "createSurface": {
      "surfaceId": "button-action",
      "catalogId": "https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json"
    }
  },
  {
    "version": "1.0",
    "updateDataModel": {
      "surfaceId": "button-action",
      "path": "/status",
      "value": "Waiting for a click"
    }
  },
  {
    "version": "1.0",
    "updateComponents": {
      "surfaceId": "button-action",
      "components": [
        {
          "id": "root",
          "component": "Column",
          "children": [
            "title",
            "status",
            "action_button"
          ],
          "justify": "center",
          "align": "center"
        },
        {
          "id": "title",
          "component": "Text",
          "text": "Click the button below",
          "variant": "body"
        },
        {
          "id": "status",
          "component": "Text",
          "text": {
            "@path": "/status"
          },
          "variant": "caption"
        },
        {
          "id": "action_button",
          "component": "Button",
          "child": "button_label",
          "variant": "primary",
          "action": {
            "event": {
              "name": "button_clicked",
              "context": {}
            }
          }
        },
        {
          "id": "button_label",
          "component": "Text",
          "text": "Click me"
        }
      ]
    }
  }
]);

globalThis.actionData = undefined;
surface.onAction(action => {
  globalThis.actionData = action;
  surface.send({
    version: '1.0',
    updateDataModel: {
      surfaceId: 'button-action',
      path: '/status',
      value: `Received ${action.action.name} from ${action.action.sourceComponentId}`
    }
  });
});
display(surface);
```
