# Login form with validation

Run the first cell, enter an email and a password of at least eight characters, then click **Sign in**. The action context contains the email, not the password. This is a demonstration, not authentication.

```javascript
const { A2UISurface } = await import('/files/a2ui/kernel.js');
globalThis.surface = new A2UISurface();
surface.send([
  {
    "version": "1.0",
    "createSurface": {
      "surfaceId": "login-form",
      "catalogId": "https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json"
    }
  },
  {
    "version": "1.0",
    "updateDataModel": {
      "surfaceId": "login-form",
      "value": {
        "email": "",
        "password": "",
        "status": "Enter your details to sign in"
      }
    }
  },
  {
    "version": "1.0",
    "updateComponents": {
      "surfaceId": "login-form",
      "components": [
        {
          "id": "root",
          "component": "Card",
          "child": "main_column"
        },
        {
          "id": "main_column",
          "component": "Column",
          "children": [
            "title",
            "email_field",
            "password_field",
            "login_btn",
            "status"
          ]
        },
        {
          "id": "title",
          "component": "Text",
          "text": "## Welcome back"
        },
        {
          "id": "email_field",
          "component": "TextField",
          "label": "Email",
          "value": {
            "@path": "/email"
          },
          "@checks": [
            {
              "condition": {
                "@call": "required",
                "args": {
                  "value": {
                    "@path": "/email"
                  }
                }
              },
              "message": "Email is required"
            },
            {
              "condition": {
                "@call": "email",
                "args": {
                  "value": {
                    "@path": "/email"
                  }
                }
              },
              "message": "Please enter a valid email address"
            }
          ]
        },
        {
          "id": "password_field",
          "component": "TextField",
          "label": "Password",
          "variant": "obscured",
          "value": {
            "@path": "/password"
          },
          "@checks": [
            {
              "condition": {
                "@call": "required",
                "args": {
                  "value": {
                    "@path": "/password"
                  }
                }
              },
              "message": "Password is required"
            },
            {
              "condition": {
                "@call": "length",
                "args": {
                  "value": {
                    "@path": "/password"
                  },
                  "min": 8
                }
              },
              "message": "Password must be at least 8 characters long"
            }
          ]
        },
        {
          "id": "login_btn",
          "component": "Button",
          "child": "login_btn_text",
          "variant": "primary",
          "@checks": [
            {
              "condition": {
                "@call": "and",
                "args": {
                  "values": [
                    {
                      "@call": "email",
                      "args": {
                        "value": {
                          "@path": "/email"
                        }
                      }
                    },
                    {
                      "@call": "length",
                      "args": {
                        "value": {
                          "@path": "/password"
                        },
                        "min": 8
                      }
                    }
                  ]
                }
              },
              "message": "Please fix errors before signing in"
            }
          ],
          "action": {
            "event": {
              "name": "login",
              "context": {
                "email": {
                  "@path": "/email"
                }
              }
            }
          }
        },
        {
          "id": "login_btn_text",
          "component": "Text",
          "text": "Sign in"
        },
        {
          "id": "status",
          "component": "Text",
          "text": {
            "@path": "/status"
          },
          "variant": "caption"
        }
      ]
    }
  }
]);

globalThis.actionData = undefined;
surface.onAction(action => {
  globalThis.actionData = action;
  surface.send({
    version: '1.0',
    updateDataModel: {
      surfaceId: 'login-form',
      path: '/status',
      value: `Signed in as ${action.action.context.email}`
    }
  });
});
display(surface);
```
