# A2UI Jupyter manual smoke test

Run the form cell, enter a name, and click **Submit**. The callback replaces the form with a confirmation. Run the last cell to inspect the action.

```javascript
const { A2UISurface } = await import('/files/a2ui/kernel.js');
globalThis.surface = new A2UISurface();
surface.send([
  {
    "version": "1.0",
    "createSurface": {
      "surfaceId": "manual-smoke-test",
      "catalogId": "https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json"
    }
  },
  {
    "version": "1.0",
    "updateDataModel": {
      "surfaceId": "manual-smoke-test",
      "path": "/person/name",
      "value": ""
    }
  },
  {
    "version": "1.0",
    "updateDataModel": {
      "surfaceId": "manual-smoke-test",
      "path": "/status",
      "value": "Waiting for submission"
    }
  },
  {
    "version": "1.0",
    "updateComponents": {
      "surfaceId": "manual-smoke-test",
      "components": [
        {
          "id": "root",
          "component": "Column",
          "children": [
            "title",
            "name",
            "submit",
            "status"
          ],
          "align": "stretch"
        },
        {
          "id": "title",
          "component": "Text",
          "text": "A2UI JupyterLab smoke test"
        },
        {
          "id": "name",
          "component": "TextField",
          "label": "Name",
          "placeholder": "Enter a name",
          "value": {
            "@path": "/person/name"
          }
        },
        {
          "id": "submit",
          "component": "Button",
          "child": "submit_label",
          "variant": "primary",
          "action": {
            "event": {
              "name": "submit",
              "context": {
                "name": {
                  "@path": "/person/name"
                }
              }
            }
          }
        },
        {
          "id": "submit_label",
          "component": "Text",
          "text": "Submit"
        },
        {
          "id": "status",
          "component": "Text",
          "text": {
            "@path": "/status"
          }
        }
      ]
    }
  }
]);
globalThis.actionData = undefined;
surface.onAction(action => {
  globalThis.actionData = action;
  surface.send([
    {version: '1.0', updateDataModel: {
      surfaceId: 'manual-smoke-test', path: '/status', value: `Submitted: ${action.action.context.name}`
    }},
    {version: '1.0', updateComponents: {surfaceId: 'manual-smoke-test', components: [
      {id: 'root', component: 'Column', children: ['title', 'status'], align: 'stretch'}
    ]}}
  ]);
});
display(surface);
```
