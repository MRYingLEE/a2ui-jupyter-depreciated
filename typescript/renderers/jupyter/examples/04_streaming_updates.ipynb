# Streaming updates

Run the first cell to render an empty list. The next cell appends restaurants one at a time; the final cell edits a restaurant and adds a footer.

```javascript
const { A2UISurface } = await import('/files/a2ui/kernel.js');
globalThis.surface = new A2UISurface();
surface.send([
  {
    "version": "1.0",
    "createSurface": {
      "surfaceId": "streaming-updates",
      "catalogId": "https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json"
    }
  },
  {
    "version": "1.0",
    "updateDataModel": {
      "surfaceId": "streaming-updates",
      "value": {
        "restaurants": []
      }
    }
  },
  {
    "version": "1.0",
    "updateComponents": {
      "surfaceId": "streaming-updates",
      "components": [
        {
          "id": "root",
          "component": "Column",
          "children": [
            "heading",
            "restaurant_list"
          ]
        },
        {
          "id": "heading",
          "component": "Text",
          "text": "## Restaurants"
        },
        {
          "id": "restaurant_list",
          "component": "Column",
          "children": {
            "path": "/restaurants",
            "componentId": "restaurant_card"
          }
        },
        {
          "id": "restaurant_card",
          "component": "Column",
          "children": [
            "card_title",
            "card_subtitle"
          ]
        },
        {
          "id": "card_title",
          "component": "Text",
          "text": {
            "@path": "title"
          }
        },
        {
          "id": "card_subtitle",
          "component": "Text",
          "text": {
            "@path": "subtitle"
          },
          "variant": "caption"
        }
      ]
    }
  }
]);

display(surface);
```

Append one restaurant every second.

```javascript
const restaurants = [
  {title: 'The Golden Fork', subtitle: 'Fine dining and spirits'},
  {title: "Ocean's Bounty", subtitle: 'Fresh daily seafood'},
  {title: 'Pizzeria Roma', subtitle: 'Wood-fired pizza'}
];
for (const [index, restaurant] of restaurants.entries()) {
  surface.send({
    version: '1.0',
    updateDataModel: {surfaceId: 'streaming-updates', path: `/restaurants/${index}`, value: restaurant}
  });
  await new Promise(resolve => setTimeout(resolve, 1000));
}
```

Change the first subtitle and add a footer.

```javascript
surface.send([
  {
    "version": "1.0",
    "updateDataModel": {
      "surfaceId": "streaming-updates",
      "path": "/restaurants/0/subtitle",
      "value": "Fine dining, now open late"
    }
  },
  {
    "version": "1.0",
    "updateComponents": {
      "surfaceId": "streaming-updates",
      "components": [
        {
          "id": "root",
          "component": "Column",
          "children": [
            "heading",
            "restaurant_list",
            "footer"
          ]
        },
        {
          "id": "footer",
          "component": "Text",
          "text": "Updated in place by a later cell",
          "variant": "caption"
        }
      ]
    }
  }
]);
```
