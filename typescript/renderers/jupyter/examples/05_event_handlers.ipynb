# Event handlers

`onAction` handles action events and `onMessage` handles all renderer messages. Each returns an unsubscribe function. Click **Add 2** to update the counter and message log.

```javascript
const adapterUrl = globalThis.document
  ? new URL('../files/a2ui/kernel.js', document.baseURI)
  : new URL('../../../../files/a2ui/kernel.js', location.href);
const { A2UISurface } = await import(adapterUrl.href);
globalThis.surface = new A2UISurface();
surface.send([
  {
    "version": "1.0",
    "createSurface": {
      "surfaceId": "event-handlers",
      "catalogId": "https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json"
    }
  },
  {
    "version": "1.0",
    "updateDataModel": {
      "surfaceId": "event-handlers",
      "value": {
        "count": 0,
        "log": "No messages yet"
      }
    }
  },
  {
    "version": "1.0",
    "updateComponents": {
      "surfaceId": "event-handlers",
      "components": [
        {
          "id": "root",
          "component": "Column",
          "children": [
            "title",
            "count_text",
            "add_button",
            "log_text"
          ],
          "align": "center"
        },
        {
          "id": "title",
          "component": "Text",
          "text": "## Counter"
        },
        {
          "id": "count_text",
          "component": "Text",
          "text": {
            "@path": "/count"
          }
        },
        {
          "id": "add_button",
          "component": "Button",
          "child": "add_label",
          "variant": "primary",
          "action": {
            "event": {
              "name": "increment",
              "context": {
                "step": 2
              }
            }
          }
        },
        {
          "id": "add_label",
          "component": "Text",
          "text": "Add 2"
        },
        {
          "id": "log_text",
          "component": "Text",
          "text": {
            "@path": "/log"
          },
          "variant": "caption"
        }
      ]
    }
  }
]);

globalThis.count = 0;
globalThis.unsubscribeAction = surface.onAction(message => {
  const action = message.action;
  if (action.name !== 'increment') return;
  globalThis.count += action.context.step;
  surface.send({
    version: '1.0',
    updateDataModel: {surfaceId: 'event-handlers', path: '/count', value: globalThis.count}
  });
});
globalThis.unsubscribeMessage = surface.onMessage(message => {
  const kind = Object.keys(message).find(key => key !== 'version');
  surface.send({
    version: '1.0',
    updateDataModel: {surfaceId: 'event-handlers', path: '/log', value: `Last renderer message: ${kind}`}
  });
});
display(surface);
```

Remove the counter handler. Clicks still update the message log.

```javascript
unsubscribeAction();
console.log('Counter handler removed. The message handler is still active.');
```

Remove the message handler too.

```javascript
unsubscribeMessage();
console.log('All handlers removed.');
```
