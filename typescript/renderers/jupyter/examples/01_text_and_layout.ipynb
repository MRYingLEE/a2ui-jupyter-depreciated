# Text and layout

Run the first cell to render a card with text, a divider, and a row. Run the next cell to update the bound greeting.

```javascript
const { A2UISurface } = await import('/files/a2ui/kernel.js');
globalThis.surface = new A2UISurface();
surface.send([
  {
    "version": "1.0",
    "createSurface": {
      "surfaceId": "text-and-layout",
      "catalogId": "https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json"
    }
  },
  {
    "version": "1.0",
    "updateDataModel": {
      "surfaceId": "text-and-layout",
      "path": "/greeting",
      "value": "Hello from the data model"
    }
  },
  {
    "version": "1.0",
    "updateComponents": {
      "surfaceId": "text-and-layout",
      "components": [
        {
          "id": "root",
          "component": "Card",
          "child": "main"
        },
        {
          "id": "main",
          "component": "Column",
          "children": [
            "heading",
            "intro",
            "divider",
            "row",
            "greeting"
          ]
        },
        {
          "id": "heading",
          "component": "Text",
          "text": "## Layout basics"
        },
        {
          "id": "intro",
          "component": "Text",
          "text": "Columns stack children vertically. Rows place them side by side.",
          "variant": "body"
        },
        {
          "id": "divider",
          "component": "Divider"
        },
        {
          "id": "row",
          "component": "Row",
          "children": [
            "left",
            "right"
          ],
          "justify": "spaceBetween"
        },
        {
          "id": "left",
          "component": "Text",
          "text": "Left"
        },
        {
          "id": "right",
          "component": "Text",
          "text": "Right",
          "variant": "caption"
        },
        {
          "id": "greeting",
          "component": "Text",
          "text": {
            "@path": "/greeting"
          }
        }
      ]
    }
  }
]);

display(surface);
```

Update `/greeting` without rebuilding the surface.

```javascript
surface.send({
  version: '1.0',
  updateDataModel: {surfaceId: 'text-and-layout', path: '/greeting', value: 'Updated from a later cell'}
});
```
